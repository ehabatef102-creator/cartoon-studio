# إنتاج حلقة كرتون على سحابة جوجل المجانية (Colab)

هذا الدفتر يشغّل **مسار الإنتاج الكامل** (سيناريو + صور سينمائية + تعليق صوتي عربي + موسيقى ومؤثرات + مونتاج فيديو) **بكامل قوته على معالج جوجل السحابي** — بدون أي استخدام لطاقة جهازك.

التكلفة: صفر. الطريقة: `Runtime → Run all`.

```
# 1) تجهيز البيئة: سحب الكود وتثبيت المتطلبات
import os, sys, shutil
if not os.path.isdir('/content/cartoon-studio'):
    os.system('git clone --depth 1 https://github.com/ehabatef102-creator/cartoon-studio.git')
os.chdir('/content/cartoon-studio')
os.system('pip install -q -r requirements.txt')
print('OK')
# تحقق من وجود ffmpeg (مثبت أصلاً على Colab)
print('ffmpeg:', shutil.which('ffmpeg'))
```

```
# 2) الإعدادات — عدّل حسب رغبتك
#@title قصة الحلقة (أو اتركها فارغة واستخدم سلسلة جاهزة)
STORY = "" #@param {type:"string"}
INDEX = 1 #@param {type:"integer"}  # سلسلة جاهزة من المكتبة (1-3) تُستخدم لو القصة فارغة
RENDER_VIDEO = True #@param {type:"boolean"}
MUSIC = True #@param {type:"boolean"}
MOTION = True #@param {type:"boolean"}
SEED = 42 #@param {type:"integer"}
```

```
# 3) تشغيل الإنتاج الكامل على السحابة
import subprocess
cmd = [sys.executable, 'scripts/produce.py', '--out', 'output', '--seed', str(SEED)]
if STORY.strip():
    cmd += ['--story', STORY.strip()]
else:
    cmd += ['--index', str(INDEX)]
if RENDER_VIDEO:
    cmd += ['--video']
if MUSIC:
    cmd += ['--music']
if MOTION:
    cmd += ['--motion']
print('RUN:', ' '.join(cmd))
proc = subprocess.run(cmd)
print('exit code:', proc.returncode)
```

```
# 4) تحميل الناتج (ZIP يحتوي السيناريو + الصور + الأصوات + الفيديو)
import glob, os, zipfile
from google.colab import files
zips = sorted(glob.glob('output/*.zip'))
print('الملفات الناتجة:')
for z in zips:
    print(' -', z)
if zips:
    files.download(zips[-1])
else:
    print('لا يوجد ZIP بعد — راجع الخطوة 3.')
```
